### Refactored version

```python
def compute_sum(a: float, b: float) -> float:
    """
    Compute the sum of two numbers.

    Args:
        a (float): The first operand.
        b (float): The second operand.

    Returns:
        float: The sum of a and b.
    """
    # ← Comment this line: explain that we are computing the result by adding a and b
    result = a + b
    # ← Comment this line: explain that we are printing the result for user‐feedback
    print("Sum is", result)
    return result


def main() -> None:
    """
    Execute the main workflow: define two numbers and compute their sum.
    """
    # ← Comment this line: defining the first number
    first_number = 10.0
    # ← Comment this line: defining the second number
    second_number = 20.0
    # ← Comment this line: calling the compute_sum function with descriptive variable names
    compute_sum(first_number, second_number)


if __name__ == "__main__":
    # ← Comment this line: ensure main() only runs when script is executed directly
    main()
```
