# 1. Correlation matrix

**a.** Compute correlation matrix of the three quantitative variables on year $1952$, $1987$ and then $2007$ using `pd.corr()`. Give a brief description of each correlation matrix.

```python
# To do
# Load the dataset into a pandas DataFrame
df_gapminder = gapminder

# Create dataframes for each year
df_1952 = df_gapminder[df_gapminder['year'] == 1952]
df_1987 = df_gapminder[df_gapminder['year'] == 1987]
df_2007 = df_gapminder[df_gapminder['year'] == 2007]

# Compute correlation matrices using pd.corr()
corr_1952 = df_1952[['lifeExp', 'pop', 'gdpPercap']].corr()
corr_1987 = df_1987[['lifeExp', 'pop', 'gdpPercap']].corr()
corr_2007 = df_2007[['lifeExp', 'pop', 'gdpPercap']].corr()

# Display the correlation matrices
print("Correlation Matrix for 1952:\n", corr_1952)
print("\nCorrelation Matrix for 1987:\n", corr_1987)
print("\nCorrelation Matrix for 2007:\n", corr_2007)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Set a theme for better aesthetics
sns.set_theme()

# Create scatter plots for each year
plt.figure(figsize=(15, 5))

# Scatter plot for 1952
plt.subplot(1, 3, 1)
sns.scatterplot(data=df_1952, x='gdpPercap', y='lifeExp')
plt.title('1952: GDP vs Life Expectancy')
plt.xscale('log')  # Log scale for better visualization if GDP values are widely spread
plt.xlabel('GDP per Capita')
plt.ylabel('Life Expectancy')

# Scatter plot for 1987
plt.subplot(1, 3, 2)
sns.scatterplot(data=df_1987, x='gdpPercap', y='lifeExp')
plt.title('1987: GDP vs Life Expectancy')
plt.xscale('log')
plt.xlabel('GDP per Capita')
plt.ylabel('Life Expectancy')

# Scatter plot for 2007
plt.subplot(1, 3, 3)
sns.scatterplot(data=df_2007, x='gdpPercap', y='lifeExp')
plt.title('2007: GDP vs Life Expectancy')
plt.xscale('log')
plt.xlabel('GDP per Capita')
plt.ylabel('Life Expectancy')

# Display the plots
plt.tight_layout()
plt.show()
```

**c.** We will try to confirm this using Analysis of Variance (ANOVA). 

In 2007,

- Perform one way ANOVA on `lifeExp` and `continent` (you can use `f_oneway` from `scipy.stats` module).

```python
import pandas as pd
from scipy.stats import f_oneway

# Filter the dataset for the year 2007
df_2007 = df_gapminder[df_gapminder['year'] == 2007]

# Group by continent and collect life expectancy values
groups = [group['lifeExp'].values for name, group in df_2007.groupby('continent')]

# Perform one-way ANOVA
anova_result = f_oneway(*groups)

# Output the ANOVA result
print(f"F-statistic: {anova_result.statistic:.4f}")
print(f"P-value: {anova_result.pvalue:.4f}")
```

- Do the same with `gdpPercap` and `continent`.

```python
import pandas as pd
from scipy.stats import f_oneway

# Filter the dataset for the year 2007
df_2007 = df_gapminder[df_gapminder['year'] == 2007]

# Group by continent and collect GDP per capita values
gdp_groups = [group['gdpPercap'].values for name, group in df_2007.groupby('continent')]

# Perform one-way ANOVA
anova_gdp_result = f_oneway(*gdp_groups)

# Output the ANOVA result
print(f"F-statistic: {anova_gdp_result.statistic:.4f}")
print(f"P-value: {anova_gdp_result.pvalue:.4f}")
```

# 3. Quntitative vs Qualitative

a. We have seen how life expectancy and economy vary across continents in 2007 (Slide 25 of the course). Compute 
-squared correlation between continent and lifeExp, then continent with gdpPercap in 1952, 1987 and 2007. \\

Do you find the results reasonable?

```python
def eta_squared(df_gapminder, cat_col, num_col):
    # Compute means per category
    mean_total = df_gapminder[num_col].mean()
    ss_total = ((df_gapminder[num_col] - mean_total)**2).sum()
    ss_between = df_gapminder.groupby(cat_col).apply(lambda g: len(g) * (g[num_col].mean() - mean_total)**2).sum()
    return ss_between / ss_total

# ---- Life Expectancy (2007)
eta2_life_2007 = eta_squared(df_gapminder[df_gapminder['year']==2007], 'continent', 'lifeExp')

# ---- GDP per capita (1952, 1987, 2007)
eta2_gdp_1952 = eta_squared(df_gapminder[df_gapminder['year']==1952], 'continent', 'gdpPercap')
eta2_gdp_1987 = eta_squared(df_gapminder[df_gapminder['year']==1987], 'continent', 'gdpPercap')
eta2_gdp_2007 = eta_squared(df_gapminder[df_gapminder['year']==2007], 'continent', 'gdpPercap')

eta2_life_2007, eta2_gdp_1952, eta2_gdp_1987, eta2_gdp_2007
```

# 3. Time evolution

We have looked at the world on three frames so far (1952, 1987 and 2007). Now, we will summarize the world from 1952 to 2007 in one graph using animation tool from `plotly.express`.

- `Seaborn` was built on `Matplotlib`, making it simpler for statistical plots and integrating well with `Pandas`. It is great for quick, aesthetic and informative statistical plots.
- `Plotly` on the other hand built as highly interactive plots with features like zooming, panning, and hover information. It works well in web applications and dashboards. There are extensive customization options for aesthetics, and it supports 3D visualizations, complex plots and animations. Read more here: [https://plotly.com/python/](https://plotly.com/python/).

**a.** Using `plotly`, create one scatterplot that summarizes the world using all information: `gdpPercap`, `lifeExp`, `pop`, `continent` and set option **animation_frame="year"** which will create frame-by-frame animated scatterplot of the world from 1952 to 2007. 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, spearmanr
# Function for eta-squared (categorical vs numeric)
def eta_squared(df_gapminder, cat_col, num_col):
    mean_total = df_gapminder[num_col].mean()
    ss_total = ((df_gapminder[num_col] - mean_total)**2).sum()
    ss_between = df_gapminder.groupby(cat_col).apply(lambda g: len(g) * (g[num_col].mean() - mean_total)**2).sum()
    return ss_between / ss_total

# Containers
years = sorted(df_gapminder['year'].unique())
pearson_corr, spearman_corr, eta2_life, eta2_gdp = [], [], [], []

for y in years:
    d = df_gapminder[df_gapminder['year'] == y]
    
    # Pearson & Spearman between lifeExp and gdpPercap
    p_corr, _ = pearsonr(d['lifeExp'], d['gdpPercap'])
    s_corr, _ = spearmanr(d['lifeExp'], d['gdpPercap'])
    
    # η² between continent and numeric vars
    e2_life = eta_squared(d, 'continent', 'lifeExp')
    e2_gdp  = eta_squared(d, 'continent', 'gdpPercap')
    
    pearson_corr.append(p_corr)
    spearman_corr.append(s_corr)
    eta2_life.append(e2_life)
    eta2_gdp.append(e2_gdp)

# Combine to DataFrame
evolution = pd.DataFrame({
    'year': years,
    'Pearson(lifeExp, GDP)': pearson_corr,
    'Spearman(lifeExp, GDP)': spearman_corr,
    'η²(continent→lifeExp)': eta2_life,
    'η²(continent→GDP)': eta2_gdp
})

print(evolution.round(3))
```

```python
plt.figure(figsize=(10,6))
plt.plot(evolution['year'], evolution['Pearson(lifeExp, GDP)'], label='Pearson(lifeExp, GDP)')
plt.plot(evolution['year'], evolution['Spearman(lifeExp, GDP)'], label='Spearman(lifeExp, GDP)')
plt.plot(evolution['year'], evolution['η²(continent→lifeExp)'], label='η²(continent→lifeExp)')
plt.plot(evolution['year'], evolution['η²(continent→GDP)'], label='η²(continent→GDP)')

plt.title("Evolution of Correlations (1952–2007)")
plt.xlabel("Year")
plt.ylabel("Correlation / η²")
plt.legend()
plt.grid(True)
plt.show()
```
